```
# 1. Instalación de librerías necesarias
!pip install py3Dmol requests -q

import requests
import py3Dmol

# Secuencia basada en el nombre ANGEL DAVID RIOS
sequence_raw = "ANGELDAVIDRIOS"

# Reemplazo de 'O' por 'K' (Lisina) para compatibilidad con el predictor biológico
sequence_clean = sequence_raw.replace("O", "K")

print(f"Generando estructura 3D para la secuencia: {sequence_clean}...")

# 2. Petición de predicción de estructura a la API de ESMFold (Meta AI)
url = "https://api.esmatlas.com/foldSequence/v1/pdb/"
response = requests.post(url, data=sequence_clean)

if response.status_code == 200:
    pdb_data = response.text

    # Guardar archivo de coordenadas PDB
    with open("angel_david_rios.pdb", "w") as f:
        f.write(pdb_data)

    print("¡Estructura predecida con éxito!")

    # 3. Visualización interactiva en 3D
    view = py3Dmol.view(width=700, height=500)
    view.addModel(pdb_data, "pdb")

    # Estilo de representación: Cinta (Cartoon) + Átomos (Sticks)
    view.setStyle({'cartoon': {'color': 'spectrum'}})
    view.addStyle({'stick': {'radius': 0.15, 'colorscheme': 'carbon'}})

    view.zoomTo()
    view.show()
else:
    print(f"Error al conectar con la API de ESMFold. Código de estado: {response.status_code}")
```
